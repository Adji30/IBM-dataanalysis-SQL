```python
import pandas as pd
import sqlite3

# 1. Connect to the local SQLite database
con = sqlite3.connect("RealWorldData.db")

# 2. Load the datasets directly into SQLite
print("Loading data...")
url_schools = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DB0201EN-SkillsNetwork/labs/FinalModule_Coursera_V5/data/ChicagoPublicSchools.csv"
df_schools = pd.read_csv(url_schools)
df_schools.to_sql("SCHOOLS", con, if_exists='replace', index=False, method="multi")

url_census = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DB0201EN-SkillsNetwork/labs/FinalModule_Coursera_V5/data/ChicagoCensusData.csv"
df_census = pd.read_csv(url_census)
df_census.to_sql("CENSUS_DATA", con, if_exists='replace', index=False, method="multi")
print("Data loaded successfully.\n")

# 3. Define the SQL Query (Problem 11/12 example)
query = """
SELECT CD.hardship_index
FROM CENSUS_DATA CD, SCHOOLS CPS
WHERE CD.community_area_number = CPS.community_area_number
AND CPS.college_enrollment = 4368;
"""

# 4. Execute the query and display the result using Pandas
print("Executing Query:")
result_df = pd.read_sql_query(query, con)
display(result_df)
```

### Query the database system catalog to retrieve table metadata

##### You can verify that the table creation was successful by retrieving the list of all tables in your schema and checking whether the SCHOOLS table was created


```python
# type in your query to retrieve list of all tables in the database for your db2 schema (username)
query_cek_tabel = """
SELECT name
FROM sqlite_master
WHERE type='table' AND name='SCHOOLS';
"""

print("Mengecek keberadaan tabel SCHOOLS:")
hasil_cek = pd.read_sql_query(query_cek_tabel, con)
display(hasil_cek)
```

### Query the database system catalog to retrieve column metadata

##### The SCHOOLS table contains a large number of columns. How many columns does this table have?


```python
# type in your query to retrieve the number of columns in the SCHOOLS table
# Kueri khusus SQLite untuk menghitung jumlah kolom di tabel SCHOOLS
query_jumlah_kolom = """
SELECT count(*) AS jumlah_kolom
FROM pragma_table_info('SCHOOLS');
"""

print("Menghitung jumlah kolom di tabel SCHOOLS:")
hasil_kolom = pd.read_sql_query(query_jumlah_kolom, con)
display(hasil_kolom)
```

Double-click **here** for a hint

<!--
In Db2 the system catalog table called SYSCAT.COLUMNS contains the column metadata
-->


Double-click **here** for the solution.

<!-- Solution:

%sql select count(*) from SYSCAT.COLUMNS where TABNAME = 'SCHOOLS'

-->


Now retrieve the the list of columns in SCHOOLS table and their column type (datatype) and length.


```python
# type in your query to retrieve all column names in the SCHOOLS table along with their datatypes and length
# Kueri SQLite untuk mengambil nama kolom dan tipe datanya
query_detail_kolom = """
SELECT name AS nama_kolom, type AS tipe_data
FROM pragma_table_info('SCHOOLS');
"""

print("Daftar Kolom dan Tipe Data di Tabel SCHOOLS:")
hasil_detail = pd.read_sql_query(query_detail_kolom, con)

# Mengatur agar Pandas menampilkan semua baris (karena ada 78 kolom)
pd.set_option('display.max_rows', None)
display(hasil_detail)

# Mengembalikan pengaturan tampilan Pandas ke normal (opsional)
pd.reset_option('display.max_rows')
```

### Problem 3

##### Which schools have highest Safety Score?

### Problem 6

### Problem 7

##### Which Schools have Average Student Attendance lower than 70%?


```python
query_prob7 = """
SELECT Name_of_School, Average_Student_Attendance
FROM SCHOOLS
WHERE CAST(REPLACE(Average_Student_Attendance, '%', '') AS REAL) < 70
ORDER BY Average_Student_Attendance;
"""

print("Menjalankan Problem 7:")
hasil_prob7 = pd.read_sql_query(query_prob7, con)
display(hasil_prob7)
```

Double-click **here** for a hint

<!--
The datatype of the "Average_Student_Attendance" column is varchar.
So you cannot use it as is in the where clause for a numeric comparison.
First use the CAST() function to cast it as a DECIMAL or DOUBLE
e.g. CAST("Column_Name" as DOUBLE)
or simply: DECIMAL("Column_Name")
-->


Double-click **here** for another hint

<!--
Don't forget the '%' age sign needs to be removed before casting
-->


Double-click **here** for the solution.

<!-- Solution:

or,
